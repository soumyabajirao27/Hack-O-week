```python
def greet(name):
    """Prints a personalized greeting"""
    print(f"Hello, {name}!")

greet("John") 

def add(x, y):
    """Returns the sum of two numbers"""
    return x + y

print(add(5, 3)) 


def greet_with_message(name, message="Hello"):
    """Prints a personalized greeting with a message"""
    print(f"{message}, {name}!")

greet_with_message("John") 
greet_with_message("John", "Hi")  
```

```python
class Car:
    def __init__(self, brand, model, year):
        """Initializes a new Car object"""
        self.brand = brand
        self.model = model
        self.year = year
        self.mileage = 0

    def drive(self, miles):
        """Updates the mileage of the car"""
        self.mileage += miles

    def describe_car(self):
        """Prints a description of the car"""
        print(f"This car is a {self.year} {self.brand} {self.model} with {self.mileage} miles.")

my_car = Car("Toyota", "Camry", 2020)
my_car.describe_car() 
my_car.drive(100)
my_car.describe_car()
```
